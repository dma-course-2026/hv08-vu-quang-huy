## Bài 6 - Diện tích khu vườn

**Yêu cầu:** Tính diện tích hình chữ nhật còn lại sau khi xây hình tròn bán kính a/2.

```python
def remaining_garden_area(a, b, pi=3.14):
    rectangle_area = a * b
    circle_area = pi * (a / 2) ** 2
    return rectangle_area - circle_area

print(f"{remaining_garden_area(10, 20):.2f}")
```

## Bài 9 - Tính tiền điện

**Yêu cầu:** Tính tiền điện theo ba bậc: 60 kW đầu giá 5, 100 kW tiếp theo giá 8, phần còn lại giá 10.

```python
def electricity_bill(previous, current):
    usage = current - previous
    if usage < 0:
        raise ValueError("chỉ số mới không được nhỏ hơn chỉ số cũ")
    first = min(usage, 60)
    second = min(max(usage - 60, 0), 100)
    third = max(usage - 160, 0)
    return first * 5 + second * 8 + third * 10

print(f"{electricity_bill(1500, 1580):.2f}")
```

## Bài 10 - Đọc số có 3 chữ số

**Yêu cầu:** Tách hàng trăm, hàng chục và hàng đơn vị.

```python
def split_three_digit_number(n):
    if not 100 <= n <= 999:
        raise ValueError("n phải có 3 chữ số")
    return n // 100, (n % 100) // 10, n % 10

print(split_three_digit_number(456))
```

## Bài 14 - Hình tam giác

**Yêu cầu:** Kiểm tra tam giác và tính chu vi, diện tích bằng công thức Heron.

```python
import math

def triangle_measurements(a, b, c):
    if min(a, b, c) <= 0 or a + b <= c or a + c <= b or b + c <= a:
        raise ValueError("ba cạnh không tạo thành tam giác")
    perimeter = a + b + c
    semi_perimeter = perimeter / 2
    area = math.sqrt(semi_perimeter * (semi_perimeter - a) * (semi_perimeter - b) * (semi_perimeter - c))
    return perimeter, area

print(triangle_measurements(3, 4, 5))
```

## Bài 16 - Tam giác cân

**Yêu cầu:** Kiểm tra tam giác có ít nhất hai cạnh bằng nhau.

```python
def is_isosceles_triangle(a, b, c):
    valid = min(a, b, c) > 0 and a + b > c and a + c > b and b + c > a
    return valid and (a == b or b == c or a == c)

print(is_isosceles_triangle(5, 5, 8))
```

## Bài 17 - Tam giác vuông

**Yêu cầu:** Kiểm tra tam giác vuông bằng định lý Pythagoras.

```python
def is_right_triangle(a, b, c):
    sides = sorted((a, b, c))
    return sides[0] > 0 and sides[0] ** 2 + sides[1] ** 2 == sides[2] ** 2

print(is_right_triangle(3, 4, 5))
```

## Bài 20 - Phương trình bậc 2

**Yêu cầu:** Giải phương trình ax^2 + bx + c = 0 theo delta.

```python
import math

def solve_quadratic_equation(a, b, c):
    if a == 0:
        return solve_linear_equation(b, c)
    delta = b ** 2 - 4 * a * c
    if delta < 0:
        return "vô nghiệm"
    if delta == 0:
        return (-b / (2 * a),)
    root = math.sqrt(delta)
    return ((-b + root) / (2 * a), (-b - root) / (2 * a))

print(solve_quadratic_equation(1, -3, 2))
```

## Bài 28 - Thực hiện phép toán

**Yêu cầu:** Thực hiện phép toán +, -, *, / và báo lỗi cho phép toán không hợp lệ hoặc chia cho 0.

```python
def calculate(a, b, operator):
    if operator == "+":
        return a + b
    if operator == "-":
        return a - b
    if operator == "*":
        return a * b
    if operator == "/":
        if b == 0:
            raise ValueError("không thể chia cho 0")
        return a / b
    raise ValueError("phép toán không hợp lệ")

print(calculate(10, 5, "+"))
```

## Bài 29 - Số ngày của năm

**Yêu cầu:** Xác định năm thường hay năm nhuận.

```python
def days_in_year(year):
    leap_year = year % 400 == 0 or (year % 4 == 0 and year % 100 != 0)
    return 366 if leap_year else 365

print(days_in_year(2024))
```

## Bài 30 - Đổi giờ phút giây

**Yêu cầu:** Đổi tổng số giây thành giờ, phút, giây.

```python
def seconds_to_hms(total_seconds):
    if total_seconds < 0:
        raise ValueError("số giây không được âm")
    hours, remainder = divmod(total_seconds, 3600)
    minutes, seconds = divmod(remainder, 60)
    return hours, minutes, seconds

print(seconds_to_hms(3665))
```

## Bài 31 - Đồng hồ điện tử

**Yêu cầu:** Tăng thời điểm h:p:s lên một giây và quay về 0 sau 24 giờ.

```python
def next_second(hour, minute, second):
    total = (hour * 3600 + minute * 60 + second + 1) % 86400
    return seconds_to_hms(total)

print(next_second(23, 59, 58))
```

## Bài 32 - Hệ phương trình bậc nhất hai ẩn

**Yêu cầu:** Giải hệ ax + by = m và cx + dy = n bằng định thức.

```python
def solve_two_by_two(a, b, m, c, d, n):
    determinant = a * d - b * c
    determinant_x = m * d - b * n
    determinant_y = a * n - m * c
    if determinant != 0:
        return determinant_x / determinant, determinant_y / determinant
    if determinant_x == 0 and determinant_y == 0:
        return "vô số nghiệm"
    return "vô nghiệm"

print(solve_two_by_two(2, 3, 8, 4, -1, -1))
```

## Bài 33 - Điểm thuộc đường tròn

**Yêu cầu:** Phân loại điểm A nằm trong, trên hay ngoài đường tròn.

```python
import math

def circle_point_position(x0, y0, radius, xa, ya, tolerance=1e-9):
    distance = math.hypot(xa - x0, ya - y0)
    if math.isclose(distance, radius, abs_tol=tolerance):
        return "trên đường tròn"
    return "trong đường tròn" if distance < radius else "ngoài đường tròn"

print(circle_point_position(0, 0, 5, 3, 4))
```

## Bài 35 - Tổng từ 1 đến n

**Yêu cầu:** Tính tổng 1 + 2 + ... + n bằng vòng lặp.

## Bài 38 - Giai thừa n!

**Yêu cầu:** Tính n! bằng vòng lặp.

```python
def factorial(n):
    if n < 0:
        raise ValueError("n phải không âm")
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

print(factorial(5))
```

## Bài 39 - Lũy thừa a^n

**Yêu cầu:** Tính a^n bằng vòng lặp.

```python
def power(base, exponent):
    if exponent < 0:
        raise ValueError("số mũ phải không âm")
    result = 1
    for _ in range(exponent):
        result *= base
    return result

print(power(2, 10))
```

## Bài 40 - Tổng số chia hết cho 3 hoặc 5

**Yêu cầu:** Tính tổng các số trong [M, N] chia hết cho 3 hoặc 5.

```python
def sum_multiples_of_3_or_5(start, end):
    return sum(i for i in range(start, end + 1) if i % 3 == 0 or i % 5 == 0)

print(sum_multiples_of_3_or_5(10, 20))
```
