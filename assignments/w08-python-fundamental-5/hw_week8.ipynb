## Bài 9

```python
from datetime import datetime

ngay_sinh=input("Nhập ngày sinh (yyyy/mm/dd): ")
try:
    sinh=datetime.strptime(ngay_sinh,"%Y/%m/%d").date()
    hom_nay=datetime.now().date()
    if sinh>hom_nay:
        print("Ngày sinh không hợp lệ")
    else:
        tuoi=hom_nay.year-sinh.year
        if (hom_nay.month,hom_nay.day)<(sinh.month,sinh.day):
            tuoi-=1
        print("Tuổi:",tuoi)
except ValueError:
    print("Ngày sinh không hợp lệ")
```

## Bài 22

```python
so_du=0
print("Nhập D số_tiền hoặc W số_tiền; Enter trống để kết thúc")
while True:
    s=input("Giao dịch: ").strip()
    if s=="":
        break
    a=s.split()
    if len(a)!=2:
        print("Giao dịch không hợp lệ")
        continue
    loai=a[0].upper()
    try:
        tien=int(a[1])
    except ValueError:
        print("Số tiền không hợp lệ")
        continue
    if tien<0:
        print("Số tiền phải không âm")
    elif loai=="D":
        so_du+=tien
    elif loai=="W":
        so_du-=tien
    else:
        print("Chỉ nhận D hoặc W")
print("Số dư:",so_du)
```

## Bài 23

```python
def kiem_tra_doi_xung(a):
    return a==a[::-1]

def sx_doi_xung(a):
    b=a.copy()
    for i in range(len(b)//2):
        b[-i-1]=b[i]
    return b

s=input("Nhập dãy số nguyên cách nhau bởi dấu phẩy: ")
a=[int(x.strip()) for x in s.split(",")]
if kiem_tra_doi_xung(a):
    print("Dãy đã đối xứng:",a)
else:
    print("Dãy ban đầu chưa đối xứng")
    b=sx_doi_xung(a)
    print("Dãy sau khi biến đổi:",b)
    print("Kiểm tra đối xứng:",kiem_tra_doi_xung(b))
```
