## Bài 51

```python
def bai_51_b(so_tien, menh_gia_1, menh_gia_2, menh_gia_3):
    if so_tien<0 or min(menh_gia_1, menh_gia_2, menh_gia_3)<=0:
        return "Số tiền phải không âm và các mệnh giá phải dương."

    ket_qua=[f"Loai {menh_gia_1} | Loai {menh_gia_2} | Loai {menh_gia_3}"]
    for so_to_1 in range(so_tien//menh_gia_1+1):
        for so_to_2 in range((so_tien-so_to_1*menh_gia_1)//menh_gia_2+1):
            con_lai=so_tien-so_to_1*menh_gia_1-so_to_2*menh_gia_2
            if con_lai%menh_gia_3==0:
                so_to_3=con_lai//menh_gia_3
                ket_qua.append(f"{so_to_1} {so_to_2} {so_to_3}")
    return "\n".join(ket_qua) if len(ket_qua)>1 else "Không có cách đổi tiền phù hợp."

so_tien=int(input("Nhập số tiền cần đổi: "))
menh_gia_1=int(input("Nhập loại tiền thứ 1: "))
menh_gia_2=int(input("Nhập loại tiền thứ 2: "))
menh_gia_3=int(input("Nhập loại tiền thứ 3: "))
print(bai_51_b(so_tien, menh_gia_1, menh_gia_2, menh_gia_3))
```

## Bài 57

```python
def sinh_fibonacci():
    a, b=0, 1
    while True:
        yield a
        a, b=b, a+b

def bai_57(n):
    if n<0:
        return "n phải không âm."
    if n==0:
        return "Số lượng số yêu cầu phải lớn hơn 0."
        
    fibo_gen=sinh_fibonacci()
    
    danh_sach_fibo=[next(fibo_gen) for _ in range(n)]
    
    ket_qua=" ".join(map(str, danh_sach_fibo))
    return f"{n} số đầu tiên của dãy số Fibonacci:\n{ket_qua}"

n=int(input("Nhập số lượng số Fibonacci n: "))
print(bai_57(n))
```

## Bài 59

```python
def sang_nguyen_to(gioi_han):
    if gioi_han<=2:
        return []
        
    la_nguyen_to=[True]*gioi_han
    la_nguyen_to[0]=la_nguyen_to[1]=False
    
    for i in range(2, int(gioi_han**0.5)+1):
        if la_nguyen_to[i]:
            for j in range(i*i, gioi_han, i):
                la_nguyen_to[j]=False
                
    return [so for so in range(2, gioi_han) if la_nguyen_to[so]]
    
def bai_59(n):
    if n<=2:
        return f"Tất cả các số nguyên tố nhỏ hơn {n} là: "
        
    # Gọi hàm bổ trợ để lấy danh sách số nguyên tố
    ket_qua=sang_nguyen_to(n)
    
    return f"Tất cả các số nguyên tố nhỏ hơn {n} là: "+" ".join(map(str, ket_qua))

n=int(input("Nhập số nguyên dương n = "))
print(bai_59(n))
```
