## Bài 5

```python
def so_sanh_chuoi(s1,s2):
    if len(s1)>len(s2):
        print(s1)
    elif len(s1)<len(s2):
        print(s2)
    else:
        print(s1)
        print(s2)

s1=input("Nhập chuỗi thứ nhất: ")
s2=input("Nhập chuỗi thứ hai: ")
so_sanh_chuoi(s1,s2)
```

## Bài 27

```python
class Circle:
    def __init__(self,ban_kinh):
        self.ban_kinh=ban_kinh

    def area(self):
        return 3.14*self.ban_kinh**2

r=float(input("Nhập bán kính: "))
hinh_tron=Circle(r)
print("Diện tích:",hinh_tron.area())
```

## Bài 28

```python
class Hinhchunhat:
    def __init__(self,chieu_dai,chieu_rong):
        self.chieu_dai=chieu_dai
        self.chieu_rong=chieu_rong

    def area(self):
        return self.chieu_dai*self.chieu_rong

a=float(input("Nhập chiều dài: "))
b=float(input("Nhập chiều rộng: "))
hinh=Hinhchunhat(a,b)
print("Diện tích:",hinh.area())
```

## Bài 29

```python
class Shape:
    def area(self):
        print(0)

class Square(Shape):
    def __init__(self,canh):
        self.canh=canh

    def area(self):
        print(self.canh**2)

hinh=Shape()
hinh.area()
hinh_vuong=Square(5)
hinh_vuong.area()
```
