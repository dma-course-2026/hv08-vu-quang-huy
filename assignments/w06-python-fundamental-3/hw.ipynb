## Bài 9. Tính tiền điện

```python
def tinhtiendien():
    ho_ten = input("Họ tên: ")
    chiso1, chiso2 = map(int, input("Chỉ số điện cũ, mới: ").split())

    so_dien = chiso2 - chiso1

    if so_dien <= 60:
        tien_dien = so_dien*5
    elif so_dien <= 160:
        tien_dien = 60*5 + (so_dien - 60)*8
    else:
        tien_dien = 60*5 + 100*8 + (so_dien - 160)*10

    print("Họ tên:", ho_ten)
    print(f"Tiền điện: {tien_dien:.0f} đồng")

tinhtiendien()
```

## Bài 14. Hình tam giác

```python
def tamgiac():
    import math

    a, b, c = map(float, input("a, b, c = ").split())

    if a > 0 and b > 0 and c > 0 and a + b > c and a + c > b and b + c > a:
        p = (a + b + c) / 2
        dien_tich = math.sqrt(p*(p - a)*(p - b)*(p - c))
        print("Chu vi:", a+b+c)
        print("Diện tích:", dien_tich)
    else:
        print("Không phải tam giác")

tamgiac()
```

## Bài 20. Phương trình bậc 2

```python
def phuongtrinhbac2():
    import math

    a = float(input("a = "))
    b = float(input("b = "))
    c = float(input("c = "))

    if a == 0:
        if b == 0:
            if c == 0:
                print("Vô số nghiệm")
            else:
                print("Vô nghiệm")
        else:
            print("x =", -c/b)
    else:
        delta = b**2 - 4*a*c

        if delta < 0:
            print("Vô nghiệm")
        elif delta == 0:
            print("Nghiệm kép:", -b/(2*a))
        else:
            x1 = (-b + math.sqrt(delta)) / (2*a)
            x2 = (-b - math.sqrt(delta)) / (2*a)
            print("Hai nghiệm:")
            print("x1 =", x1)
            print("x2 =", x2)

phuongtrinhbac2()
```

## Bài 28. Máy tính 2 ngôi

```python
def maytinh2ngo():
    a = int(input("Số thứ 1: "))
    b = int(input("Số thứ 2: "))
    phep_toan = input("Phép toán (+, -,*, /): ")

    if phep_toan == "+":
        ket_qua = a + b
    elif phep_toan == "-":
        ket_qua = a - b
    elif phep_toan == "*":
        ket_qua = a*b
    elif phep_toan == "/":
        if b == 0:
            print("Không chia được cho 0")
            ket_qua = None
        else:
            ket_qua = a / b
    else:
        print("Phép toán không hợp lệ")
        ket_qua = None

    if ket_qua is not None:
        print(f"{a} {phep_toan} {b} = {ket_qua}")

maytinh2ngo()
```

## Bài 32. Hệ phương trình bậc 1 hai ẩn

```python
def hephuongtrinh():
    a = float(input("a = "))
    b = float(input("b = "))
    m = float(input("m = "))
    c = float(input("Nhập c = "))
    d = float(input("Nhập d = "))
    n = float(input("Nhập n = "))

    D = a*d - b*c
    Dx = m*d - b*n
    Dy = a*n - c*m

    if D != 0:
        x = Dx / D
        y = Dy / D
        print("Có 1 nghiệm:")
        print("x =", x, "và y =", y)
    elif Dx == 0 and Dy == 0:
        print("Vô số nghiệm")
    else:
        print("Vô nghiệm")

hephuongtrinh()
```

## Bài 33. Điểm thuộc đường tròn

```python
def diemthuocduongtron():
    import math

    R = float(input("Nhập bán kính R: "))
    x0 = float(input("x0 = "))
    y0 = float(input("y0 = "))
    xa = float(input("xa = "))
    ya = float(input("ya = "))

    d = math.sqrt((xa - x0)**2 + (ya - y0)**2)

    if d < R:
        print("A nằm trong đường tròn")
    elif d == R:
        print("A nằm trên đường tròn")
    else:
        print("A nằm ngoài đường tròn")

diemthuocduongtron()
```

## Bài 45. Kiểm tra số nguyên tố

```python
def songuyento():
    n = int(input("n = "))

    if n < 2:
        la_nguyen_to = False
    else:
        la_nguyen_to = True
        for i in range(2, int(n**0.5) + 1):
            if n % i == 0:
                la_nguyen_to = False
                break

    if la_nguyen_to:
        print(n, "là số nguyên tố")
    else:
        print(n, "không phải số nguyên tố")

songuyento()
```

## Bài 47. Các số hoàn hảo nhỏ hơn n

```python
def lietkesohoanhao():
    n = int(input("n = "))

    for so in range(2, n):
        tong_uoc = 0
        for i in range(1, so):
            if so % i == 0:
                tong_uoc += i

        if tong_uoc == so:
            print(so, end=" ")
    print()

lietkesohoanhao()
```

## Bài 49. Số Armstrong có 3 chữ số

```python
def armstrong():
    for n in range(100, 1000):
        tram = n // 100
        chuc = (n // 10) % 10
        don_vi = n % 10

        if n == tram**3 + chuc**3 + don_vi**3:
            print(n)

armstrong()
```

## Bài 50. Số có tổng chữ số bằng tích chữ số

```python
def tongbangtich():
    for a in range(1, 10):
        for b in range(10):
            for c in range(10):
                if a + b + c == a*b*c:
                    print(a, b, c)

tongbangtich()
```
