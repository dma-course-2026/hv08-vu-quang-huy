## Bài 14. Hình tam giác

```python
import math

a, b, c = map(float, input("a, b, c = ").split())

if a > 0 and b > 0 and c > 0 and a + b > c and a + c > b and b + c > a:
    p = (a + b + c) / 2
    dien_tich = math.sqrt(p*(p - a)*(p - b)*(p - c))
    print("Chu vi:", a+b+c)
    print("Diện tích:", dien_tich)
else:
    print("Không phải tam giác")
```

## Bài 20. Phương trình bậc 2

```python
import math

a = float(input("a = "))
b = float(input("b = "))
c = float(input("c = "))

if a == 0:
    if b == 0:
        if c == 0:
            print("Vô số nghiệm")
        else:
            print("Vô nghiệm")
    else:
        print("x =", -c/b)
else:
    delta = b**2 - 4*a*c

    if delta < 0:
        print("Vô nghiệm")
    elif delta == 0:
        print("Nghiệm kép:", -b/(2*a))
    else:
        x1 = (-b + math.sqrt(delta)) / (2*a)
        x2 = (-b - math.sqrt(delta)) / (2*a)
        print("Hai nghiệm:")
        print("x1 =", x1)
        print("x2 =", x2)
```

## Bài 32. Hệ phương trình bậc 1 hai ẩn

```python
a = float(input("a = "))
b = float(input("b = "))
m = float(input("m = "))
c = float(input("Nhập c = "))
d = float(input("Nhập d = "))
n = float(input("Nhập n = "))

D = a*d - b*c
Dx = m*d - b*n
Dy = a*n - c*m

if D != 0:
    x = Dx / D
    y = Dy / D
    print("Có 1 nghiệm:")
    print("x =", x, "và y =", y)
elif Dx == 0 and Dy == 0:
    print("Vô số nghiệm")
else:
    print("Vô nghiệm")
```

## Bài 33. Điểm thuộc đường tròn

```python
import math

R = float(input("Nhập bán kính R: "))
x0 = float(input("x0 = "))
y0 = float(input("y0 = "))
xa = float(input("xa = "))
ya = float(input("ya = "))

d = math.sqrt((xa - x0)**2 + (ya - y0)**2)

if d < R:
    print("A nằm trong đường tròn")
elif d == R:
    print("A nằm trên đường tròn")
else:
    print("A nằm ngoài đường tròn")
```

## Bài 45. Kiểm tra số nguyên tố

```python
n = int(input("n = "))

if n < 2:
    la_nguyen_to = False
else:
    la_nguyen_to = True
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            la_nguyen_to = False
            break

if la_nguyen_to:
    print(n, "là số nguyên tố")
else:
    print(n, "không phải số nguyên tố")
```
